## 1. Overview

<!-- What is this concept? 2–4 sentences that a newcomer could understand.
     Include: what problem it solves, why it exists, where it fits in the AI landscape. -->

Traditional search engines work by matching the words in your query against an index of pages. Deep Search flips that model. It uses artificial intelligence to interpret the meaning and intent behind a query, not just the words in it. Search "best restaurant" on a regular engine and you get pages that happen to contain those words. Deep Search instead tries to understand what you're actually looking for and why.

The term covers a few related implementations, each with its own emphasis. 

* IBM's version is built for large, specialized document collections: it collects, converts, and curates sources like patents and research papers into structured, searchable knowledge graphs. 

* Google's Deep Search, part of AI Mode, works more like a research assistant: it searches hundreds of websites, cross checks arguments across them, and produces a documented report in minutes. 

A closely related idea, "Deep Research," goes further still by actually reading the retrieved sources in depth and synthesizing them into a full report, rather than just surfacing relevant snippets. For an intermediate learner, these distinctions matter less than the shared mechanics, which is what the diagram below walks through.

## 2. How It Works

<!-- Break the concept into numbered steps or subsections.
     Use diagrams (images from assets/) where helpful.
     Each subsection should follow: description → danger/difficulty level → counter-technique or note -->

### 2.1 User question
You ask something in plain language, not a string of keywords.
### 2.2 Query decomposition
The system breaks your question down into multiple sub-questions, similar to how a human researcher would outline a topic before digging in.
### 2.3 Multi-source search
For each sub-question, the system fires off targeted searches. Some tools run up to 20 unique searches across web pages, academic papers, or internal documents.
### 2.4 Retrieval and extraction
The raw results are converted into a form the model can reason over. IBM's approach, for example, converts PDF documents into a structured JSON format, then applies natural language processing and computer vision to build searchable knowledge graphs. The model then ranks and pulls the passages most relevant to each sub-question.
### 2.5 Synthesis and citation
The model reads across everything it gathered, resolves conflicts or agreements between sources, and writes one coherent answer. The AI does not just find relevant articles, it reads through the sources and synthesizes findings into a structured response, sometimes with citations to the originals.

## 3. Advantages & Limitations

| | Aspect | Commentary |
|--|--------|------------|
| 🟢 | **Complex queries** | Handles multi part questions instead of single keyword lookups. |
| 🟢 | **Time saved** | Cuts down the manual work of opening and skimming many pages yourself. |
| 🟢 | **Coverage** | 	Can pull from huge, hard to search collections like patents or research papers. |
| 🟢 | **Cross checking** | Checks multiple sources at once, which helps surface disagreements or gaps. |
| 🟢 | **Output format** | Produces a structured, often cited, answer instead of a bare list of links. |
| 🔴 | **Speed and cost** | Introduces real computational cost and slower response times compared to a normal search query. |
| 🔴 | **Accuracy risk** | There is a risk of the model generating inaccurate information despite the extra steps. |
| 🔴 | **Source dependency** | Quality depends heavily on the sources it can reach. If access is restricted, the answer is incomplete. |
| 🔴 | **Smoothing over conflicts** | The synthesis step can blur genuine disagreements between sources instead of flagging them clearly. |
| 🔴 | **Auditability** | 	Harder to audit than a simple ranked list, since the reasoning happens inside the model. |

## 4. Key Takeaways
<div style="font-size: 16px; line-height: 1.6;">

- **Deep Search understands intent, not just keywords.**
- **Query decomposition is the first and most important step.**
- **Multiple parallel searches replace one single lookup.**
- **Synthesis turns scattered sources into one coherent, cited answer.**
- **The extra reasoning brings real costs: latency, compute, and risk of errors.**
