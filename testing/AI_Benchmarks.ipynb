> <span style="font-size:20px;">This notebook serves to document benchmark used to compare/test AI models. </span>


| Benchmark | Category | What it tests | How it tests |
|---|---|---|---|
| **MMLU** | General knowledge | Broad academic knowledge across domains | Multiple-choice questions across ~57 university-level subjects (law, medicine, physics, history...) |
| **MMLU-Pro** | General knowledge | Harder version of MMLU | More difficult questions, more answer options to reduce lucky guesses |
| **GPQA (Diamond)** | General knowledge | PhD-level science | Highly specialized questions, difficult even with web search access |
| **BIG-Bench Hard (BBH)** | Reasoning | Complex reasoning tasks | A subset of tasks selected for their difficulty within BIG-Bench |
| **MATH** | Mathematics | High-school/university-level math problems | Problem solving with an exact, verifiable answer |
| **AIME** | Mathematics | Competition-level math | Questions drawn from a real U.S. math competition |
| **GSM8K** | Mathematics | Elementary/middle-school-level math | Word problems requiring multi-step reasoning |
| **HumanEval** | Coding | Simple function generation | The AI writes code from a description, automatically tested |
| **SWE-bench (Verified)** | Coding | Fixing real bugs | Bug fixes on real GitHub repositories, validated by the project's own tests |
| **LiveCodeBench** | Coding | Coding on recent problems | Uses recent problems to avoid training-data contamination |
| **DeepSWE** | Coding | Fixing real software bugs | Bug-fix tasks on real projects, similar to SWE-bench |
| **FrontierSWE** | Coding | Advanced software engineering | Complex, cutting-edge-difficulty coding tasks |
| **SWE Marathon** | Coding | Long, complex coding tasks | Extended, multi-step development assignments |
| **Terminal Bench 2.1** | Coding | Command-line usage | The AI must complete system tasks via a terminal |
| **Program Bench** | Coding | Program generation/solving | Similar to HumanEval, programming exercises |
| **Chatbot Arena (LMSYS)** | Conversation | Perceived response quality | Humans blind-vote between two responses (Elo-style ranking) |
| **MT-Bench** | Conversation | Multi-turn conversations | Judged by an AI "judge" on coherence and quality |
| **AlpacaEval** | Conversation | Instruction following | Automated comparison of response quality to given instructions |
| **MMMU** | Multimodal | Understanding complex images | Questions requiring reading diagrams, charts, schematics |
| **MathVista** | Multimodal | Math with visual support | Math problems accompanied by images/charts |
| **CharXiv (RQ) w/ tool** | Visual agents | Understanding scientific charts | Questions on charts pulled from real arXiv papers, with tool use (zoom, code) |
| **Zerobench w/ tool (Pass@5)** | Visual agents | Very hard vision tasks | Questions deliberately near-impossible; 5 attempts, counted as a pass if any one succeeds |
| **WebArena** | Agents | Autonomous web navigation | The AI must complete concrete tasks on real websites |
| **GAIA** | Agents | General multi-step tasks | Combines research, reasoning, and tool use |
| **τ-bench (tau-bench)** | Agents | Realistic customer/agent interactions | Simulates customer-service scenarios with tool use |
| **BrowseComp** | Agents | Complex web research | Finding hard-to-locate information via multi-step searches |
| **Automation Bench** | Agents / office work | Digital task automation | Chaining actions such as filling out forms |
| **GDPval-AA v2 (Elo)** | Agents / office work | Realistic professional tasks | Real deliverables (reports, spreadsheets, slides) judged via pairwise comparisons scored as Elo, with terminal/web access |
| **JobBench** | Agents / office work | Occupation-specific tasks | Simulates work tied to a specific profession |
| **AA-Briefcase (Elo)** | Agents / office work | Complex office tasks | Pairwise comparisons scored as Elo (by Artificial Analysis) |
| **SpreadsheetBench 2** | Agents / office work | Spreadsheet manipulation | Creating/editing formulas and formatting in Excel/Sheets |
| **TruthfulQA** | Safety/robustness | Resistance to common false beliefs | Trick questions where a popular but false answer is tempting |
| **HellaSwag** | Safety/robustness | Common sense and narrative coherence | Choosing the most plausible continuation of a given situation |
